# SUMMARY

## I. TỔNG KẾT DỰ ÁN

### 1. Các phát hiện chính

**Câu hỏi 1: Tác động của Nhà tài trợ và Giám sát**
- **Ưu thế của khối Công nghiệp:** Nhà tài trợ **Industry** có tỷ lệ thành công cao hơn đáng kể so với nhóm Academic/Research. Điều này phản ánh quy trình tuyển chọn dự án khắt khe và nguồn lực dồi dào của các công ty dược.
- **Vai trò của DMC:** Sự hiện diện của **Hội đồng Giám sát An toàn Dữ liệu (DMC)** làm tăng tỷ lệ thành công. Cơ chế giám sát chặt chẽ giúp chuẩn hóa quy trình và giảm thiểu sai sót.
- **Quy mô không đảm bảo thành công:** Quy mô thử nghiệm lớn (số lượng người tham gia đông) không đồng nghĩa với tỷ lệ thành công cao hơn, thậm chí còn tiềm ẩn rủi ro quản lý.
- **Tương tác Sponsor-Size:** Với các thử nghiệm quy mô nhỏ, sự chênh lệch hiệu quả giữa Industry và Academic là không đáng kể, nhưng khoảng cách nới rộng ở các thử nghiệm quy mô lớn.
- **Insight chính:** Chất lượng quản trị và giám sát (Governance) là yếu tố then chốt. Sự đầu tư bài bản về quy trình của Industry và vai trò của DMC là những mô hình cần nhân rộng.

**Câu hỏi 2: Thời gian và Hiện tượng Fail Fast**
- **Nghịch lý thời gian:** Quan sát thấy thử nghiệm thành công thường kéo dài hơn, nhưng đây chủ yếu là do **Survivorship Bias** (thử nghiệm thất bại thường bị dừng sớm).
- **Quy luật 12-24 tháng:** Phần lớn các thử nghiệm thất bại bộc lộ vấn đề ngay trong **12 đến 24 tháng đầu tiên**.
- **Sàng lọc tại Phase 2:** Phase 2 có tỷ lệ thất bại cao nhất, đóng vai trò là "cửa ải" quan trọng nhất để ngăn chặn các ứng viên thuốc kém chất lượng đi tiếp.
- **Hiệu ứng Pha:** Thử nghiệm Phase 3 có tỷ lệ thành công cao nhất do chất lượng đầu vào đã được chọn lọc kỹ từ Phase 2.
- **Insight chính:** Chiến lược **"Fail Fast"** là cần thiết. Cần tập trung giám sát chặt chẽ trong 2 năm đầu để phát hiện sớm dấu hiệu thất bại, tối ưu hóa nguồn lực.

**Câu hỏi 3: Thiết kế Thử nghiệm và Chất lượng Dữ liệu**
- **Khoảng trống thông tin:** Một lượng lớn thử nghiệm thiếu thông tin (UNKNOWN) về các yếu tố thiết kế quan trọng như phương pháp phân bổ hay làm mù.
- **Cảnh báo đỏ từ Unknown:** Các thử nghiệm có giá trị **Unknown** ở các biến thiết kế có tỷ lệ thất bại cao hơn hẳn. Đây là chỉ dấu (proxy) mạnh mẽ cho chất lượng thiết kế kém.
- **Hiệu quả của thiết kế chuẩn:** Các thiết kế tiêu chuẩn vàng như **Randomized** và **Factorial** cho thấy tỷ lệ thành công cao hơn so với thiết kế đơn nhóm (Single Group).
- **Tác động khiêm tốn:** Mặc dù có ý nghĩa thống kê, nhưng sự khác biệt về tỷ lệ thành công giữa các loại hình thiết kế (nếu đã loại bỏ nhóm Unknown) là không quá lớn.
- **Insight chính:** Sự minh bạch và hoàn thiện trong hồ sơ thiết kế (**Design Completeness**) quan trọng hơn bản thân loại hình thiết kế. Thiếu thông tin là một dấu hiệu rủi ro lớn.

**Câu hỏi 4: Độ phức tạp và Tính khả thi**
- **Nghịch lý phức tạp:** Thử nghiệm càng phức tạp (nhiều điều kiện bệnh lý, nhiều can thiệp) thì tỷ lệ thành công càng có xu hướng thấp.
- **Yếu tố nhiễu từ bệnh nặng:** Việc nhắm đến **4+ điều kiện bệnh lý** thường đi kèm với các ca bệnh ung thư giai đoạn cuối/di căn, dẫn đến tỷ lệ thất bại cao do bản chất bệnh lý chứ không hẳn do thiết kế.
- **Khác biệt theo loại can thiệp:** Thử nghiệm về **Thiết bị (Device)** và Thủ thuật có tỷ lệ thành công cao hơn so với Thuốc (Drug) và Sinh học (Biological).
- **Thách thức của phối hợp thuốc:** Các thử nghiệm đa can thiệp (Combination therapy) đối mặt với rủi ro vận hành và tương tác thuốc cao hơn.
- **Insight chính:** Đơn giản hóa thiết kế (**Simplicity**) giúp tăng khả năng hoàn thành. Cần thận trọng khi diễn giải thất bại của các thử nghiệm phức tạp vì có thể do đặc điểm bệnh lý nặng.

**Câu hỏi 5: Xu hướng và Động lực Công nghệ**
- **Không có suy thoái dài hạn:** Bác bỏ giả thuyết ngành R&D ung thư đang rơi vào suy thoái. Tỷ lệ thành công biến động nhưng không giảm dần đều theo thời gian.
- **Sóng công nghệ:** Các đợt tăng trưởng tỷ lệ thành công trùng khớp với sự bùng nổ của các công nghệ mới như **Liệu pháp miễn dịch (Immunotherapy)** giai đoạn 2011-2014.
- **Quy mô mở rộng:** Số lượng thử nghiệm lâm sàng tăng trưởng liên tục, cho thấy dòng vốn và sự quan tâm của xã hội với lĩnh vực này vẫn rất lớn.
- **Khả năng phục hồi:** Ngành có khả năng phục hồi nhanh chóng sau các cú sốc ngoại cảnh như đại dịch COVID-19.
- **Insight chính:** Đổi mới công nghệ (**Innovation**) là động lực chính phá vỡ các rào cản nghiên cứu. Đầu tư vào các nền tảng công nghệ mới là hướng đi đúng đắn nhất.

**Câu hỏi : Machine Learning - Dự báo thất bại thử nghiệm lâm sàng ung thư**
- **Mô hình tốt nhất:** XGBoost (Tuned) đạt ROC-AUC cao nhất (~0.86)
- **Yếu tố quan trọng nhất:**
  1. Enrollment Per Arm (số lượng tuyển dụng mỗi nhánh)
  2. Enrollment Count (tổng quy mô tuyển dụng)
  3. Industry Sponsor (nhà tài trợ doanh nghiệp)
  4. Late Phase (giai đoạn thử nghiệm muộn)

- **Insight chính:** Quy mô tuyển dụng và giai đoạn thử nghiệm là các chỉ báo mạnh mẽ nhất; các dự án có nguồn lực lớn và ở giai đoạn sau có xác suất thành công cao hơn đáng kể

### 2. Hạn chế của Đồ án

- **Dữ liệu thiếu hụt (Missing Data):** Tỷ lệ giá trị thiếu (Unknown/Null) khá cao ở các biến quan trọng như thiết kế thử nghiệm, đặc điểm nhân khẩu học, gây khó khăn cho việc phân tích sâu.
- **Thiên kiến kẻ sống sót (Survivorship Bias):** Phân tích thời gian chịu ảnh hưởng nặng nề bởi việc các thử nghiệm thất bại thường kết thúc sớm, cần các phương pháp thống kê nâng cao để khắc phục.
- **Giới hạn phạm vi nguồn:** Dữ liệu chỉ lấy từ ClinicalTrials.gov, có thể chưa bao phủ hết các thử nghiệm ở các khu vực khác hoặc không được đăng ký công khai.
- **Thiếu thông tin chi tiết:** Không có dữ liệu về chi phí, mức độ tuân thủ điều trị của bệnh nhân hay các chỉ số sinh học cụ thể, làm hạn chế khả năng giải thích nguyên nhân thất bại ở cấp độ vi mô.
- **Đơn giản hóa biến số:** Một số biến phức tạp (như loại can thiệp, bệnh lý) phải được gộp nhóm hoặc đơn giản hóa để đưa vào mô hình, có thể làm mất mát thông tin.

### 3. Hướng phát triển

- **Mở rộng nguồn dữ liệu:** Kết hợp thêm dữ liệu từ các nguồn khác (như PubMed, FDA reports, Patent databases) để có cái nhìn đa chiều hơn.
- **Ứng dụng NLP nâng cao:** Sử dụng các mô hình ngôn ngữ lớn (LLM) để trích xuất thông tin chi tiết hơn từ các trường văn bản tự do (như tiêu chí lựa chọn, mô tả chi tiết), giúp điền khuyết dữ liệu thiếu.
- **Phân tích chuỗi thời gian sâu:** Áp dụng các mô hình Time Series tiên tiến để dự báo xu hướng R&D trong tương lai chính xác hơn.
- **Mô hình hóa rủi ro đa biến:** Xây dựng hệ thống chấm điểm rủi ro (Risk Scoring System) tổng hợp từ tất cả các yếu tố (Sponsor, Design, Complexity, Trend) để hỗ trợ ra quyết định sớm.
- **Nghiên cứu định tính:** Kết hợp với nghiên cứu định tính (phỏng vấn chuyên gia) để lý giải sâu hơn các phát hiện định lượng (ví dụ: tại sao thiết kế Single Group lại thất bại nhiều ở một số chỉ định cụ thể).

### 23120348 - Ngô Thị Thục Quyên 

#### Thách thức và Khó khăn

- **Thách thức kỹ thuật:** Dữ liệu thử nghiệm lâm sàng có quy mô lớn, nhiều biến định tính, tỷ lệ giá trị thiếu cao và phân phối lệch mạnh, đặc biệt ở các biến phản ánh quy mô và độ phức tạp. Việc duy trì tính nhất quán giữa các bước tiền xử lý và tạo đặc trưng cũng là một thách thức đáng kể.
- **Cách vượt qua:** Các quyết định tiền xử lý được đưa ra dựa trên phân tích dữ liệu khám phá, kết hợp giữa tiêu chí thống kê và ý nghĩa lĩnh vực. Giá trị ngoại lai được xử lý linh hoạt (biến đổi log, phân nhóm) thay vì loại bỏ trực tiếp. Quy trình fit-transform được tách biệt rõ ràng để tránh rò rỉ dữ liệu.
- **Thách thức lớn nhất:** Việc cân bằng giữa tính chặt chẽ về mặt phương pháp dữ liệu và tính hợp lý trong bối cảnh nghiên cứu lâm sàng.
#### Học hỏi và Phát triển

- **Kỹ năng kỹ thuật:** Kỹ năng xử lý dữ liệu thực tế, xây dựng pipeline tạo đặc trưng và chuẩn hóa dữ liệu theo chuẩn Machine Learning được củng cố rõ rệt.
- **Phương pháp phân tích:** Tầm quan trọng của phân tích khám phá dữ liệu trong việc định hướng các quyết định tiền xử lý và tạo đặc trưng được làm rõ.
- **Kiến thức lĩnh vực:** Hiểu biết sâu hơn về cách quy mô, độ phức tạp và thiết kế nghiên cứu phản ánh thực tế vận hành của thử nghiệm lâm sàng.
- **Điều gây ngạc nhiên:** Nhiều yếu tố có ý nghĩa thống kê nhưng mức độ ảnh hưởng thực tế lại ở mức thấp, cho thấy không tồn tại yếu tố đơn lẻ mang tính quyết định.
- **Định hình hiểu biết:** Dự án này giúp làm rõ rằng Khoa học Dữ liệu không chỉ là kỹ thuật phân tích, mà là quá trình chuyển dữ liệu thành insight có ý nghĩa để hỗ trợ quyết định thực tế.
